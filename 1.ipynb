```python
# --------------------------------------------------
# 1. Check actual data types
# --------------------------------------------------

print("S1 dtypes")
display(s1.dtypes)

print("\nS2 dtypes")
display(s2.dtypes)

print("\nS3 dtypes")
display(s3.dtypes)


# --------------------------------------------------
# 2. Ground-truth match count
# --------------------------------------------------

gt_analysis = ground_truth.copy()

gt_analysis["matched_entity_ids"] = (
    gt_analysis["matched_entity_ids"]
    .fillna("")
    .astype(str)
)

gt_analysis["num_matches"] = gt_analysis["matched_entity_ids"].apply(
    lambda x: 0 if x.strip() == "" else len(x.split(","))
)

print("\nMatch-count statistics:")
display(gt_analysis["num_matches"].describe())

print("\nMatch-count distribution:")
display(
    gt_analysis["num_matches"]
    .value_counts()
    .sort_index()
    .head(30)
)

print(
    "\nSingleton S1 entities:",
    (gt_analysis["num_matches"] == 0).sum()
)

print(
    "S1 entities with at least one match:",
    (gt_analysis["num_matches"] > 0).sum()
)

print(
    "S1 entities with multiple matches:",
    (gt_analysis["num_matches"] > 1).sum()
)


# --------------------------------------------------
# 3. Look at actual examples
# --------------------------------------------------

pd.set_option("display.max_colwidth", 150)

print("\nSource 1 examples")
display(s1.sample(10, random_state=42))

print("\nSource 2 examples")
display(s2.sample(10, random_state=42))

print("\nSource 3 examples")
display(s3.sample(10, random_state=42))


# --------------------------------------------------
# 4. Memory usage
# --------------------------------------------------

def memory_mb(df):
    return df.memory_usage(deep=True).sum() / 1024**2

print("\nMemory:")
print(f"S1: {memory_mb(s1):.2f} MB")
print(f"S2: {memory_mb(s2):.2f} MB")
print(f"S3: {memory_mb(s3):.2f} MB")
print(f"GT: {memory_mb(ground_truth):.2f} MB")
```

```python
import pandas as pd

# ---------------------------------------------------------
# Select some S1 entities with different match counts
# ---------------------------------------------------------

samples = []

for count, n, seed in [
    (0, 5, 10),   # singleton / no match
    (1, 5, 20),   # exactly one match
    (3, 5, 30),   # three matches
    (7, 5, 40),   # many matches
]:
    ids = (
        gt_analysis.loc[
            gt_analysis["num_matches"] == count,
            "source1_entity_id"
        ]
        .sample(n=min(n, (gt_analysis["num_matches"] == count).sum()),
                random_state=seed)
        .tolist()
    )
    samples.extend(ids)

# ---------------------------------------------------------
# Ground-truth rows for selected S1 entities
# ---------------------------------------------------------

sample_gt = ground_truth[
    ground_truth["source1_entity_id"].isin(samples)
].copy()

# ---------------------------------------------------------
# Collect all matched S2/S3 IDs
# ---------------------------------------------------------

s2_ids = set()
s3_ids = set()

for _, row in sample_gt.iterrows():
    matched = str(row["matched_entity_ids"])

    if matched == "" or matched == "nan":
        continue

    for entity_id in matched.split(","):
        entity_id = entity_id.strip()

        if entity_id.startswith("S2-"):
            s2_ids.add(entity_id)

        elif entity_id.startswith("S3-"):
            s3_ids.add(entity_id)

# ---------------------------------------------------------
# Fetch only these records from the huge datasets
# ---------------------------------------------------------

sample_s1 = s1[
    s1["entity_id"].isin(samples)
].copy()

sample_s2 = s2[
    s2["entity_id"].isin(s2_ids)
].copy()

sample_s3 = s3[
    s3["entity_id"].isin(s3_ids)
].copy()

# ---------------------------------------------------------
# Display
# ---------------------------------------------------------

for _, row in sample_gt.iterrows():

    s1_id = row["source1_entity_id"]
    matches = str(row["matched_entity_ids"])

    print("=" * 100)
    print("S1:", s1_id)
    print("Ground Truth:", matches)

    print("\nSOURCE 1")
    display(
        sample_s1[
            sample_s1["entity_id"] == s1_id
        ]
    )

    matched_ids = []

    if matches != "" and matches != "nan":
        matched_ids = [
            x.strip() for x in matches.split(",")
        ]

    if matched_ids:
        print("\nMATCHED SOURCE 2")
        display(
            sample_s2[
                sample_s2["entity_id"].isin(matched_ids)
            ]
        )

        print("\nMATCHED SOURCE 3")
        display(
            sample_s3[
                sample_s3["entity_id"].isin(matched_ids)
            ]
        )
```
